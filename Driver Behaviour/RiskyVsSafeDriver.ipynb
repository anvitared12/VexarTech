Why i am not considering speed and max_speed individually is because avg_speed hides the time when driver actually drove at high speed and max_speed me sometimes flag a person who actually drove at high speed during highway which is not risky.

### Flagging risky drivers using telemetry data


```python
telemetry.head()
```

Longitude and Latitude does not matter to flagging drivers as speed is already given.

Below i have taken threshold data from internet to flag each accelerator value. and then added all the values and stored in Accel_flag column
<p>Internet Data:<p>
1. Lateral Movement (Accel_X_g) — Safe at < 0.5gYes, this is completely safe. Normal, smooth lane changes and relaxed cornering usually stay between 0.1g and 0.3g.Staying below 0.5g means the rider is not swerving erratically or cutting sharply through traffic.

2. Acceleration (Accel_Y_g Positive) — Safe at < 0.5gYes, this is safe. Accelerating up to 0.4g is standard for pulling away from traffic lights smoothly.Only when they cross +0.5g does it mean they are snapping the throttle open aggressively or lifting the front wheel.

3. Braking (Accel_Y_g Negative) — Caution ZoneBe careful here: While a value like -0.3g or -0.4g is safe, normal braking, a value of -0.5g is already getting close to a very firm stop.For braking, you generally want to see values less than 0.55g or 0.6g before you flag them. So, while -0.4g is safe, -0.5g is right on the edge of a harsh brake.

```python
x_flag = (
    (telemetry["Accel_X_g"] > 0.5) |
    (telemetry["Accel_X_g"] < -0.5)
).astype(int)

y_flag = (
    (telemetry["Accel_Y_g"] > 0.4) |
    (telemetry["Accel_Y_g"] < -0.6)
).astype(int)

z_flag = (
    (telemetry["Accel_Z_g"] > 1.5) |
    (telemetry["Accel_Z_g"] < 0.5)
).astype(int)

# Sum all three flags
telemetry["Accel_flag"] = x_flag + y_flag + z_flag
```

```python
telemetry.head()
```

Same with case of Gyrometer.
<p> Internet Data:</p>
<p>
1. Gyro_X_dps (Roll Rate - Lean Speed):Flag Level: > ± 40°/sec to 50°/secWhy: This measures how quickly the rider drops the bike into a lean or pulls it back up. While a high lean angle is normal, a high lean rate (flick rate) means they are aggressively cutting corners, racing, or performing quick "zig-zag" maneuvers through traffic. </p>
<p>
2. Gyro_Y_dps (Pitch Rate - Wheelies & Stoppies):Flag Level: > ± 25°/secWhy: This measures the bike's nose dipping or lifting. High values paired with rapid deceleration indicate a "stoppie" (rear wheel off the ground). High values paired with rapid acceleration indicate a wheelie.</p>
<p>
3. Gyro_Z_dps (Yaw Rate - Heading Rotations):Flag Level: > ± 35°/secWhy: This tracks the steering rotation. Sharp U-turns or slow maneuvers natively cause high yaw, but a massive spike at high speeds points directly to a dangerous skid, fish-tailing, or a low-side crash.</P>

```python
gx_flag = (
    (telemetry["Gyro_X_dps"] > 40) |
    (telemetry["Gyro_X_dps"] < -40)
).astype(int)

gy_flag = (
    (telemetry["Gyro_Y_dps"] > 25) |
    (telemetry["Gyro_Y_dps"] < -25)
).astype(int)

gz_flag = (
    (telemetry["Gyro_Z_dps"] > 35) |
    (telemetry["Gyro_Z_dps"] < -35)
).astype(int)

telemetry["Gyro_flag"] = (
    gx_flag + gy_flag + gz_flag
)
```

At the end i have consolidated the gyrometer and accelerometer value trip wise. For this i added all the values and dived by total number of values in present per trip_ID. Which gives me good idea on how much they drove wrong from overall trips covered.

```python
def consolidate_trips(df: pd.DataFrame) -> pd.DataFrame:
    trip_df = df.groupby('Trip_ID').agg(
        Driver_ID=('Driver_ID', 'first'),
        Vehicle_ID=('Vehicle_ID', 'first') if 'Vehicle_ID' in df.columns else ('Driver_ID', 'first'),
        n_readings=('Accel_flag', 'size'),

        Accel_flag_events=('Accel_flag', lambda x: (x > 0).sum()),
        Gyro_flag_events=('Gyro_flag', lambda x: (x > 0).sum()),
    ).reset_index()

    trip_df['Accel_flag_rate'] = trip_df['Accel_flag_events'] / trip_df['n_readings']
    trip_df['Gyro_flag_rate'] = trip_df['Gyro_flag_events'] / trip_df['n_readings']

    trip_df['trip_risk_rate'] = (trip_df['Accel_flag_rate'] + trip_df['Gyro_flag_rate']) / 2

    return trip_df
```

```python
def consolidate_drivers(trip_df: pd.DataFrame) -> pd.DataFrame:
    def weighted_rate(sub, rate_col):
        return (sub[rate_col] * sub['n_readings']).sum() / sub['n_readings'].sum()

    driver_rows = []
    for driver_id, sub in trip_df.groupby('Driver_ID'):
        driver_rows.append({
            'Driver_ID': driver_id,
            'total_trips': sub['Trip_ID'].nunique(),
            'total_readings': sub['n_readings'].sum(),
            'Accel_flag_rate': weighted_rate(sub, 'Accel_flag_rate'),
            'Gyro_flag_rate': weighted_rate(sub, 'Gyro_flag_rate'),
            'avg_trip_risk_rate': weighted_rate(sub, 'trip_risk_rate'),
            'worst_trip_risk_rate': sub['trip_risk_rate'].max(),
        })

    driver_df = pd.DataFrame(driver_rows)
    return driver_df
```

I have divided into three parts for better understanding Safe, medium and risky. I took these values according the amount of rash drivings attempted by considering IMU values.

```python
def add_safety_score(driver_df: pd.DataFrame) -> pd.DataFrame:
    driver_df['safety_score'] = (1 - driver_df['avg_trip_risk_rate']) * 100
    driver_df['safety_score'] = driver_df['safety_score'].clip(0, 100).round(1)

    def tier(score):
        if score >= 85:
            return 'Safe'
        elif score >= 65:
            return 'Moderate'
        else:
            return 'Risky'

    driver_df['risk_tier'] = driver_df['safety_score'].apply(tier)
    return driver_df.sort_values('safety_score', ascending=False).reset_index(drop=True)
```

Accordingly Flagged the drivers

```python
if __name__ == "__main__":

    trip_summary = consolidate_trips(telemetry)
    driver_summary = consolidate_drivers(trip_summary)
    driver_summary = add_safety_score(driver_summary)

    print(trip_summary.head())
    print(driver_summary.head())

    trip_summary.to_csv("trip_level_summary.csv", index=False)
    driver_summary.to_csv("driver_risk_scores.csv", index=False)
```
