According to the graph every drivers avg speed is above 20.
We will defining something
if the gap between avg_speed and max_speed is very high - then he/she is a risky driver
if the gap is very minute or avg_speed > max_speed - then he/she is safe driver

```python
Trips_Telemetry_merged['speed_gap'] = Trips_Telemetry_merged['Max_Speed_kmph'] - Trips_Telemetry_merged['Avg_Speed_kmph']
Trips_Telemetry_merged['trip_risky_flag'] = (Trips_Telemetry_merged['speed_gap'] > 30).astype(int)  # threshold in km/h, adjust as needed
```

```python
Trips_Telemetry_merged.head()
```

```python
driver_risk_count = Trips_Telemetry_merged.groupby('Driver_ID_x')['trip_risky_flag'].sum().reset_index()
driver_risk_count.columns = ['Driver_ID_x', 'num_risky_trips']

# Optional: as a percentage of their total trips (fairer, since each driver has ~15 trips)
trip_count = Trips_Telemetry_merged.groupby('Driver_ID_x').size().reset_index(name='total_trips')
driver_risk_count = driver_risk_count.merge(trip_count, on='Driver_ID_x')
driver_risk_count['pct_risky_trips'] = (driver_risk_count['num_risky_trips'] / driver_risk_count['total_trips']) * 100
```

```python
driver_risk_count['category'] = driver_risk_count['pct_risky_trips'].apply(
    lambda x: 'Safe' if x == 0 else ('Medium' if x <= 20 else 'Risky')
)
```

Accel_X_g (Lateral Acceleration): Measures side-to-side movement.
High spikes: Indicate aggressive cornering, hard swerving, or sudden lane changes.
Smooth lines: Indicate gentle, controlled steering and calm driving.

Accel_Y_g (Longitudinal Acceleration): Measures forward and backward movement.
High positive values: Indicate rapid, aggressive acceleration (jackrabbit starts).
High negative values: Indicate harsh braking, tailgating behavior, or a collision.

Accel_Z_g (Vertical Acceleration): Measures up-and-down movement.
Frequent spikes: Indicate a driver who ignores speed bumps, hits potholes at high speeds, or drives recklessly on rough terrain.

```python
import numpy as np

# Accel X flag
x_flag = (
    (Trips_Telemetry_merged["Accel_X_g"] > 0.5) |
    (Trips_Telemetry_merged["Accel_X_g"] < -0.5)
).astype(int)

# Accel Y flag
y_flag = (
    (Trips_Telemetry_merged["Accel_Y_g"] > 0.4) |
    (Trips_Telemetry_merged["Accel_Y_g"] < -0.6)
).astype(int)

# Accel Z flag
z_flag = (
    (Trips_Telemetry_merged["Accel_Z_g"] > 1.5) |
    (Trips_Telemetry_merged["Accel_Z_g"] < 0.5)
).astype(int)

# Sum all three flags
Trips_Telemetry_merged["Accel_flag"] = x_flag + y_flag + z_flag
```

```python

Trips_Telemetry_merged.head()
```

```python
# Gyro X flag - Roll Rate
gx_flag = (
    (Trips_Telemetry_merged["Gyro_X_dps"] > 40) |
    (Trips_Telemetry_merged["Gyro_X_dps"] < -40)
).astype(int)

# Gyro Y flag - Pitch Rate
gy_flag = (
    (Trips_Telemetry_merged["Gyro_Y_dps"] > 25) |
    (Trips_Telemetry_merged["Gyro_Y_dps"] < -25)
).astype(int)

# Gyro Z flag - Yaw Rate
gz_flag = (
    (Trips_Telemetry_merged["Gyro_Z_dps"] > 35) |
    (Trips_Telemetry_merged["Gyro_Z_dps"] < -35)
).astype(int)

# Sum the three gyro flags
Trips_Telemetry_merged["Gyro_flag"] = (
    gx_flag + gy_flag + gz_flag
)
```

```python
Trips_Telemetry_merged.head()
```

```python
# telemetry_df: one row per minute, columns: Trip_ID, Duration_Min, Accel_flag, Gyro_flag, trip_risky_flag, speed_gap

trip_level = Trips_Telemetry_merged.groupby('Trip_ID_x').agg(
    duration_min       = ('Duration_Min', 'first'),   # same value repeated, just grab one
    trip_risky_flag     = ('trip_risky_flag', 'first'), # trip-level, same across rows
    speed_gap            = ('speed_gap', 'first'),

    accel_bad_minutes    = ('Accel_flag', lambda x: (x >= 1).sum()),   # how many minutes had ANY harsh accel
    accel_avg_severity   = ('Accel_flag', 'mean'),                      # avg intensity per minute across trip
    accel_max_severity   = ('Accel_flag', 'max'),                      # worst single minute

    gyro_bad_minutes     = ('Gyro_flag', lambda x: (x >= 1).sum()),
    gyro_avg_severity    = ('Gyro_flag', 'mean'),
    gyro_max_severity    = ('Gyro_flag', 'max'),
).reset_index()

# Normalize frequency-based columns by duration (severity columns don't need this)
trip_level['accel_bad_pct'] = (trip_level['accel_bad_minutes'] / trip_level['duration_min']) * 100
trip_level['gyro_bad_pct']  = (trip_level['gyro_bad_minutes'] / trip_level['duration_min']) * 100
```

```python
trip_level['speed_points'] = trip_level['trip_risky_flag'] * 2

trip_level['accel_points'] = (
    trip_level['accel_max_severity'] * 1.0 +          # worst moment, out of 3
    (trip_level['accel_bad_pct'] / 100) * 3            # how often it happened, scaled
)

trip_level['gyro_points'] = (
    trip_level['gyro_max_severity'] * 1.0 +
    (trip_level['gyro_bad_pct'] / 100) * 3
)

trip_level['trip_score'] = (
    trip_level['speed_points'] +
    trip_level['accel_points'] +
    trip_level['gyro_points']
)
```

```python
Trips_Telemetry_merged.head()
```

```python
driver_level = trip_level.merge(Trips_Telemetry_merged[['Trip_ID_x','Driver_ID_x']], on='Trip_ID_x').groupby('Driver_ID_x').agg(
    avg_trip_score = ('trip_score', 'mean'),
    max_trip_score = ('trip_score', 'max'),
    pct_risky_trips = ('trip_risky_flag', 'mean'),
    total_trips = ('Trip_ID_x', 'nunique')
).reset_index()

driver_level['pct_risky_trips'] *= 100

driver_level['overall_risk_rating'] = (
    (driver_level['avg_trip_score'] / driver_level['avg_trip_score'].max()) * 70 +
    (driver_level['max_trip_score'] / driver_level['max_trip_score'].max()) * 30
)

driver_level['risk_category'] = pd.cut(
    driver_level['overall_risk_rating'],
    bins=[-1, 40, 70, 100],
    labels=['Safe', 'Medium', 'Risky']
)
```
